# Search evaluation

Run the current search-quality evaluation against the golden dataset. This uses the project's current corpus, embedding configuration, and default `k=5`; it does not compare configurations.

The Make target starts the database, loads the evaluation corpus into its isolated schema, and prints the current vector, text, and hybrid search metrics.

```python
from pathlib import Path
from subprocess import run

project_root = Path.cwd().resolve()
if not (project_root / 'pyproject.toml').is_file():
    project_root = project_root.parents[1]

run(['make', 'evaluate-search'], cwd=project_root, check=True)
```

**Conclusion:** The current hybrid search is the strongest retrieval choice: it matches vector search's 98% hit rate and recall while slightly improving MRR (0.9275 versus 0.9175), for a modest 6.7 ms mean-latency increase. Full-text search is much faster but retrieves relevant evidence for only 25% of cases, so it is not suitable as the current standalone default.
