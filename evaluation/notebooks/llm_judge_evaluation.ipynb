# RAG judge evaluation

Run the current end-to-end RAG evaluation with the configured LLM judge. It assesses answer correctness against the golden reference answers; it does not compare prompts or models.

This run makes both answer-generation and judge requests for every golden case. Ensure the agent and judge OpenAI settings are available in the environment or `.env` file before running it.

```python
from pathlib import Path
from subprocess import run

project_root = Path.cwd().resolve()
if not (project_root / 'pyproject.toml').is_file():
    project_root = project_root.parents[1]

run(['make', 'evaluate-judge'], cwd=project_root, check=True)
```
